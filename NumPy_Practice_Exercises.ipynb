### Q13. Why NumPy is faster than plain Python
Create two Python lists of 500,000 numbers each. Time how long it takes to add them element-by-element using a Python loop (or list comprehension). Then convert both lists to NumPy arrays and time how long it takes to add them using `+`. Print both times and, in a comment, state which was faster and roughly by how much.

```python
import time
n = 500000
p = list(range(n))
q = list(range(n))
start = time.time()
r = [p[i] + q[i] for i in range(n)]
print('Python loop time:', round(time.time() - start, 4), 'seconds')
arr_a = np.array(p)
arr_b = np.array(q)
start = time.time()
result_array = arr_a + arr_b
print('NumPy vectorized time:', round(time.time() - start, 4), 'seconds')
```
