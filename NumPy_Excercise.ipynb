#Access Array Elements

#NumPy Array Copy vs View

```python
arr = np.array([1, 2, 3, 4, 5])
x = arr.copy()
arr[0] = 42

print(arr)
print(x)
```

```python
arr = np.array([1, 2, 3, 4, 5])
x = arr.view()
arr[0] = 42

print(arr)
print(x)
```

## 1. Hook: Python lists are too slow for ML

ML models train on datasets with millions of numbers. Let's see why plain Python loops can't keep up.

```python
import numpy as np
import time

n = 1000000
list_a = list(range(n))
list_b = list(range(n))

# Plain Python loop
start = time.time()
result_list = [list_a[i] + list_b[i] for i in range(n)]
print('Python loop time:', round(time.time() - start, 4), 'seconds')

# NumPy vectorized version
arr_a = np.array(list_a)
arr_b = np.array(list_b)

start = time.time()
result_array = arr_a + arr_b
print('NumPy vectorized time:', round(time.time() - start, 4), 'seconds')
```

**Takeaway:** NumPy is often 20–50x faster because it does the math in optimized C code instead of a Python `for` loop. This speed is what makes training ML models on large datasets practical.

## 2. Introduce the array and its shape

A single data record (one row of a dataset) is a **vector**. A whole dataset (rows × columns) is a **matrix** — a 2D NumPy array.

```python
vector = np.array([1, 2, 3])
print('1D array (vector):', vector)
print('shape:', vector.shape, '| ndim:', vector.ndim, '| dtype:', vector.dtype)

matrix = np.array([[1, 2], [3, 4], [5, 6]])
print('\n2D array (matrix) — like 3 data rows, 2 features each:')
print(matrix)
print('shape:', matrix.shape, '| ndim:', matrix.ndim)
```

A pandas DataFrame you'll use later is really a 2D NumPy array underneath, with column names attached.

## 3. Indexing and slicing

This is exactly how you will later split a dataset into **X** (features) and **y** (target).

```python
data = np.array([
    [25, 50000, 0],   # age, income, defaulted?
    [40, 80000, 0],
    [22, 20000, 1],
    [35, 60000, 0],
    [29, 15000, 1]
])

print('First row:', data[0])
print('First 3 rows:\n', data[0:3])
print('All rows, first column (age):', data[:,0])
print('All rows, last column (target y):', data[:, -1])

# This is the X / y split you'll use constantly in ML:
X = data[:, :-1]   # all columns except the last = features
y = data[:, -1]    # last column = target
print('\nX (features):\n', X)
print('y (target):', y)
```

## 4. Vectorized operations — the core idea

**Rule of thumb: if you're writing a `for` loop over data, you're probably doing it wrong.** NumPy lets you operate on an entire array at once.

```python
ages = np.array([25, 40, 22, 35, 29])

print('Ages + 1 year:', ages + 1)
print('Ages doubled:', ages * 2)
print('Ages squared:', ages ** 2)
print('Ages > 30 (boolean mask):', ages > 30)
print('Only ages > 30:', ages[ages > 30])
```
